```python
enum Color {
    Rojo,
    Verde,
    Azul,
}
```

```python
fn describir_color(color: Color) {
    match color {
        Color::Rojo => println!("Es rojo"),
        Color::Verde => println!("Es verde"),
        Color::Azul => println!("Es azul"),
    }
}
```

```python
fn dividir(a: f64, b: f64) -> Result<f64, String> {
    if b == 0.0 {
        Err(String::from("No se puede dividir por cero"))
    } else {
        Ok(a / b)
    }
}
```
